```python
# We need to convert genres in this form
# ['Action','Adventure','Fantasy','Science Fiction']
```

```python
# This module is used to convert string list to actual list
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
def convert(obj):
    l = []
    for i in ast.literal_eval(obj):
        l.append(i['name'])
    return l
```

```python
movie['genres'] = movie['genres'].apply(convert)
```

```python
# In cast data we need only top 3 actor names
```

```python
def convert3(obj):
    l = []
    count = 0
    for i in ast.literal_eval(obj):
        if count != 3:
            l.append(i['name'])
            count += 1
        else:
            break
    return l
```

```python
movie['cast'] = movie['cast'].apply(convert3)
```

```python
def fetchDirector(obj):
    l = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            l.append(i['name'])
            break
    return l
```

### Vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
vec = cv.fit_transform(df['tags']).toarray()
```

```python
vec.shape
```

```python
cv.get_feature_names_out()
```

```python
# We need to assume that love,loving,loved words are same
# To do this we need below liberary
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
ps.stem("loving"), ps.stem('loved'), ps.stem('love')
```

```python
df['tags'] = df['tags'].apply(stem)
```

```python
# We need to find diatance (angle) between two movies
# We find the cosine angle between two movies
# If angle is low similarity betwwen two movies is high
# dist -> 1/similarity
# To find this we need below liberary
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vec)
```

### Main function to recommand movie

```python
# enumerate creares tuple of index and value
list(enumerate(similarity[0]))
```

```python
# We need to sort diatances in reverse order without loosing its index
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])
```

```python
# To get top 5 similar movies
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
# Creating a function to recommand top 5 movies related to search

def recommand(movie):
    idx = df[df['title'] == movie].index[0]
    dist = similarity[idx]
    movie_list = sorted(list(enumerate(dist)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movie_list:
        print(df.iloc[i[0]].title)
```

```python
recommand('Batman Begins')
```

```python
recommand('John Carter')
```
